## 1. Lines and Slope
To graph a straight line, you essentially need two points, or a point and a slope.

### The Slope Formula
Slope ($m$) represents how a line rises or falls. It is defined as:
$$ m = \frac{\text{Rise}}{\text{Run}} = \frac{y_2 - y_1}{x_2 - x_1} $$

### Point-Slope Form
By manipulating the slope formula (fixing one point $(x_1, y_1)$ and letting the other float as $(x, y)$), we get the equation of a straight line:
$$ m = \frac{y - y_1}{x - x_1} $$
$$ y - y_1 = m(x - x_1) $$
> **Note:** This is called the **Point-Slope Form** because it requires one point and the slope to complete.

### Example 1: Finding the Equation of a Line
**Question:** Find the equation of the line passing through $(-2, -3)$ and $(8, 2)$.

*   Let $(x_1, y_1) = (-2, -3)$
*   Let $(x_2, y_2) = (8, 2)$

**Step 1: Find the slope ($m$)**
$$ m = \frac{2 - (-3)}{8 - (-2)} = \frac{5}{10} = \frac{1}{2} $$

**Step 2: Plug into Point-Slope Form**
$$ y - (-3) = \frac{1}{2}(x - (-2)) $$
$$ y + 3 = \frac{1}{2}(x + 2) $$
$$ y = \frac{1}{2}x + 1 - 3 $$
$$ y = \frac{1}{2}x - 2 $$



```python
# [TEACHING TOOL] Run this cell to display an interactive graph of the line y = 1/2x - 2
import plotly.graph_objects as go
import numpy as np

# Define x values
x = np.linspace(-10, 10, 100)
# Define y values
y = 0.5 * x - 2

fig = go.Figure()

# Add the line
fig.add_trace(go.Scatter(x=x, y=y, mode='lines', name='y = 1/2x - 2', line=dict(color='blue', width=2)))

# Add the two specific points from the example
fig.add_trace(go.Scatter(x=[-2, 8], y=[-3, 2], mode='markers', name='Given Points: (-2, -3) & (8, 2)', 
                         marker=dict(color='red', size=10, symbol='circle')))

fig.update_layout(
    title="Interactive 2D Graph: y = 1/2x - 2",
    xaxis_title="X Axis",
    yaxis_title="Y Axis",
    xaxis=dict(zeroline=True, zerolinewidth=2, zerolinecolor='black'),
    yaxis=dict(zeroline=True, zerolinewidth=2, zerolinecolor='black', scaleanchor="x", scaleratio=1),
    showlegend=True,
    template="plotly_white"
)

fig.show()
```

## 2. Slope-Intercept Form & Special Lines
When we solved for $y$ in the previous example, we arrived at:
$$ y = mx + b $$
This is the **Slope-Intercept Form**, where:
*   $m = \text{slope}$
*   $b = y\text{-intercept}$ (where the line crosses the y-axis)

### Horizontal and Vertical Lines
*   **Horizontal Lines:** $y = c$ (where $c$ is a constant). These lines have a slope of $0$ and are parallel to the x-axis.
*   **Vertical Lines:** $x = c$ (where $c$ is a constant). These lines have an undefined slope and are parallel to the y-axis.

### Example 2: Standard Form to Slope-Intercept
**Question:** Convert $4x + 2y - 3 = 0$ into slope-intercept form.
$$ 4x + 2y - 3 = 0 $$
$$ 2y = -4x + 3 $$
$$ y = -2x + \frac{3}{2} $$
*(Slope = -2, y-intercept = 1.5)*



## 3. Parallel and Perpendicular Lines
*   **Parallel Lines:** Lines that have the **same slope** ($m_1 = m_2$).
*   **Perpendicular Lines:** Lines where slopes are **negative reciprocals** of each other ($m_1 = -\frac{1}{m_2}$). They meet exactly at a $90^\circ$ angle.



```python
# [TEACHING TOOL] Interactive Graph showing Parallel and Perpendicular lines
# Let's use the line from Example 2: y = -2x + 1.5
# A Parallel line example: y = -2x + 4
# A Perpendicular line example: y = 0.5x + 1.5

x = np.linspace(-5, 5, 100)
y_orig = -2 * x + 1.5
y_para = -2 * x + 4
y_perp = 0.5 * x + 1.5

fig = go.Figure()

fig.add_trace(go.Scatter(x=x, y=y_orig, mode='lines', name='Original: y = -2x + 1.5', line=dict(color='blue')))
fig.add_trace(go.Scatter(x=x, y=y_para, mode='lines', name='Parallel: y = -2x + 4', line=dict(color='lightblue', dash='dash')))
fig.add_trace(go.Scatter(x=x, y=y_perp, mode='lines', name='Perpendicular: y = 1/2x + 1.5', line=dict(color='red')))

# Ensure aspect ratio is 1:1 so perpendicular lines visually look 90 degrees
fig.update_layout(
    title="Visualizing Parallel and Perpendicular Lines",
    xaxis_title="X Axis",
    yaxis_title="Y Axis",
    xaxis=dict(zeroline=True, zerolinewidth=2, zerolinecolor='black'),
    yaxis=dict(zeroline=True, zerolinewidth=2, zerolinecolor='black', scaleanchor="x", scaleratio=1),
    showlegend=True,
    template="plotly_white"
)

fig.show()
```

## 4. Angle of Inclination (Missed Lecture Concept)
The angle of inclination ($\theta$) is the angle a line makes with the positive x-axis.
If we visualize the rise ($\Delta y$) and run ($\Delta x$) as sides of a right triangle, we can use trigonometry:
$$ \tan(\theta) = \frac{\text{Opposite}}{\text{Adjacent}} = \frac{\Delta y}{\Delta x} $$
Since $\frac{\Delta y}{\Delta x}$ is the slope ($m$), we get this crucial relationship:
$$ m = \tan(\theta) $$

### Teaching Examples:
*   **Find slope from angle:** If $\theta = 30^\circ$ (or $\frac{\pi}{6}$ radians), then $m = \tan(30^\circ) = \frac{\sqrt{3}}{3}$.
*   **Find angle from slope:** If $m = -1$, we need to find $\theta$ where $\tan(\theta) = -1$. Using the unit circle, $\theta = 135^\circ$ (or $\frac{3\pi}{4}$ radians).

## 5. The Distance Formula (Missed Lecture Concept)
If you have two points $(x_1, y_1)$ and $(x_2, y_2)$, the distance $D$ between them is derived straight from the **Pythagorean Theorem** ($a^2 + b^2 = c^2$):
$$ D^2 = (x_2 - x_1)^2 + (y_2 - y_1)^2 $$
Taking the square root of both sides gives the Distance Formula:
$$ D = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2} $$



## 6. Bonus: Expanding to 3D Lines & Distance (Interactive 360° View)
While Calculus 1 focuses on 2D, as a teacher, you can show students how these concepts naturally extend into 3-dimensional space (Calculus 3!). 
The 3D distance formula is:
$$ D = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2 + (z_2 - z_1)^2} $$

*Run the cell below to visualize points and a line segment in 3D. **Click and drag to rotate the view 360 degrees!***



```python
# [TEACHING TOOL] Interactive 360-Degree 3D Plot using Plotly
import plotly.graph_objects as go

# Two points in 3D space
pt1 = [-2, -3, 1]
pt2 = [8, 2, 5]

x_vals = [pt1[0], pt2[0]]
y_vals = [pt1[1], pt2[1]]
z_vals = [pt1[2], pt2[2]]

fig = go.Figure()

# Plot the points
fig.add_trace(go.Scatter3d(
    x=x_vals, y=y_vals, z=z_vals,
    mode='markers+text',
    text=['Point A (-2, -3, 1)', 'Point B (8, 2, 5)'],
    textposition='bottom center',
    marker=dict(size=8, color=['blue', 'red'])
))

# Plot the line connecting them
fig.add_trace(go.Scatter3d(
    x=x_vals, y=y_vals, z=z_vals,
    mode='lines',
    line=dict(color='green', width=4),
    name="3D Line Segment"
))

fig.update_layout(
    title="3D Line and Distance (Click and Drag to Rotate 360°)",
    scene=dict(
        xaxis_title='X Axis',
        yaxis_title='Y Axis',
        zaxis_title='Z Axis'
    ),
    showlegend=False,
    template="plotly_dark"
)

fig.show()
```
