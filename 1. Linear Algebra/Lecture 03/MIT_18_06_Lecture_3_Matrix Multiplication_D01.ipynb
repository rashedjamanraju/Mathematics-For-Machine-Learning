## 1. Geometric Meaning of the Dot Product

For two vectors $\mathbf A$ and $\mathbf B$, the dot product is

$$\mathbf A\cdot\mathbf B=\|\mathbf A\|\,\|\mathbf B\|\cos\theta,$$

where $\theta$ is the angle between them.

The dot product is a **scalar**, not a vector and not itself a projection length. It combines the length of $\mathbf B$ with the **signed scalar component** of $\mathbf A$ in the direction of $\mathbf B$:

$$\mathbf A\cdot\mathbf B
=\|\mathbf B\|\left(\frac{\mathbf A\cdot\mathbf B}{\|\mathbf B\|}\right),
\qquad \mathbf B\ne\mathbf 0.$$

The signed scalar component is $\|\mathbf A\|\cos\theta$. The vector projection is a different object; both are derived in the next section.

**Projection analogy:** Imagine $\mathbf B$ as a line on the ground and $\mathbf A$ as a vector from the origin. Drop a perpendicular from the tip of $\mathbf A$ to the line spanned by $\mathbf B$. The foot of that perpendicular gives the vector projection.

- **Positive dot product:** $\theta<90^\circ$. The projection points in the same direction as $\mathbf B$.
- **Zero dot product:** $\theta=90^\circ$. The projection is the zero vector; $\mathbf A$ is perpendicular to $\mathbf B$.
- **Negative dot product:** $\theta>90^\circ$. The projection points opposite to the direction of $\mathbf B$.

The dot product measures alignment, with magnitude also depending on the lengths of the vectors.

## 2. Scalar Component vs. Vector Projection

It is important to distinguish the **signed scalar component** from the **vector projection**.

### 1. Signed scalar component

The component of $\mathbf A$ in the direction of nonzero vector $\mathbf B$ is

$$\operatorname{comp}_{\mathbf B}\mathbf A
=\|\mathbf A\|\cos\theta
=\frac{\mathbf A\cdot\mathbf B}{\|\mathbf B\|}.$$

This is a signed number:
- It is positive when the angle is acute.
- It is zero when the vectors are perpendicular.
- It is negative when the angle is obtuse.

The **unsigned length** of the projection is

$$\frac{|\mathbf A\cdot\mathbf B|}{\|\mathbf B\|}.$$

So the signed component should not always be called a length: a length cannot be negative.

### 2. Vector projection

The vector projection points along the line spanned by $\mathbf B$. Multiply the signed scalar component by the unit vector in the direction of $\mathbf B$:

$$\operatorname{proj}_{\mathbf B}\mathbf A
=\left(\frac{\mathbf A\cdot\mathbf B}{\|\mathbf B\|}\right)
\left(\frac{\mathbf B}{\|\mathbf B\|}\right)$$

Therefore,

$$\boxed{\operatorname{proj}_{\mathbf B}\mathbf A
=\frac{\mathbf A\cdot\mathbf B}{\|\mathbf B\|^2}\mathbf B},
\qquad \mathbf B\ne\mathbf 0.$$

The scalar component is a number; the vector projection has the same dimension as $\mathbf A$ and $\mathbf B$.

## 3. Visualizing Dot Product as Projection (Torchlight Analogy)

The interactive 3D plot below projects three vectors $\mathbf A_1,\mathbf A_2,\mathbf A_3$ onto the **line spanned by** $\mathbf B$ (the positive $x$-axis in this example).

- **Vector B (black):** the reference direction; it is a line direction, not a 2D ground plane.
- **A1 (blue):** has an acute angle with $\mathbf B$, so its projection points in the positive $x$ direction.
- **A2 (green):** is along the $y$-axis and perpendicular to $\mathbf B$, so its projection is the zero vector.
- **A3 (purple):** has an obtuse angle with $\mathbf B$, so its projection points in the negative $x$ direction.

The dashed yellow segments connect each vector tip to its projection foot. Each segment is perpendicular to the $x$-axis. Rotate the plot to inspect the geometry.

```python
import plotly.graph_objects as go
import numpy as np

# Define Vector B (Ground)
v_b = np.array([5, 0, 0])

# Define three different A vectors
v_a_pos = np.array([3, 0, 4])   # Acute angle -> Positive projection
v_a_zero = np.array([0, 4, 0])  # Orthogonal -> Zero projection
v_a_neg = np.array([-2, 0, 3])  # Obtuse angle -> Negative projection

# Calculate Vector Projections using the derived formula: (A . B) / ||B||^2 * B
proj_pos = (np.dot(v_a_pos, v_b) / np.linalg.norm(v_b)**2) * v_b
proj_zero = (np.dot(v_a_zero, v_b) / np.linalg.norm(v_b)**2) * v_b
proj_neg = (np.dot(v_a_neg, v_b) / np.linalg.norm(v_b)**2) * v_b

fig = go.Figure()

# 1. Plot Ground Vector B
fig.add_trace(go.Scatter3d(
    x=[0, v_b[0]], y=[0, v_b[1]], z=[0, v_b[2]],
    mode='lines+text',
    line=dict(color='black', width=6),
    text=["", "Ground Vector B"],
    textposition="bottom right",
    name="Vector B (Ground)"
))

# 2. Plot Positive Scenario
fig.add_trace(go.Scatter3d(x=[0, v_a_pos[0]], y=[0, v_a_pos[1]], z=[0, v_a_pos[2]], mode='lines+text',
    line=dict(color='blue', width=5), text=["", "Vector A1 (+ Dot)"], textposition="top center", name="A1 (Positive)"))
fig.add_trace(go.Scatter3d(x=[0, proj_pos[0]], y=[0, proj_pos[1]], z=[0, proj_pos[2]], mode='lines',
    line=dict(color='red', width=8), name="Shadow of A1"))
fig.add_trace(go.Scatter3d(x=[v_a_pos[0], proj_pos[0]], y=[v_a_pos[1], proj_pos[1]], z=[v_a_pos[2], proj_pos[2]], mode='lines',
    line=dict(color='yellow', width=3, dash='dash'), name="Torchlight 1"))

# 3. Plot Zero Scenario (Orthogonal)
fig.add_trace(go.Scatter3d(x=[0, v_a_zero[0]], y=[0, v_a_zero[1]], z=[0, v_a_zero[2]], mode='lines+text',
    line=dict(color='green', width=5), text=["", "Vector A2 (0 Dot)"], textposition="top center", name="A2 (Orthogonal)"))
# No shadow trace needed since it's 0,0,0, but we draw the torchlight straight down
fig.add_trace(go.Scatter3d(x=[v_a_zero[0], 0], y=[v_a_zero[1], 0], z=[v_a_zero[2], 0], mode='lines',
    line=dict(color='yellow', width=3, dash='dash'), name="Torchlight 2 (Drops to 0)"))

# 4. Plot Negative Scenario
fig.add_trace(go.Scatter3d(x=[0, v_a_neg[0]], y=[0, v_a_neg[1]], z=[0, v_a_neg[2]], mode='lines+text',
    line=dict(color='purple', width=5), text=["", "Vector A3 (- Dot)"], textposition="top center", name="A3 (Negative)"))
fig.add_trace(go.Scatter3d(x=[0, proj_neg[0]], y=[0, proj_neg[1]], z=[0, proj_neg[2]], mode='lines',
    line=dict(color='red', width=8), name="Shadow of A3 (Backward)"))
fig.add_trace(go.Scatter3d(x=[v_a_neg[0], proj_neg[0]], y=[v_a_neg[1], proj_neg[1]], z=[v_a_neg[2], proj_neg[2]], mode='lines',
    line=dict(color='yellow', width=3, dash='dash'), name="Torchlight 3"))


# Formatting the 360 View
fig.update_layout(
    title='Interactive 360° View: Positive, Zero, and Negative Projections',
    scene=dict(
        xaxis=dict(title='X Axis', range=[-4, 6], backgroundcolor="rgb(230, 230, 240)"),
        yaxis=dict(title='Y Axis', range=[-2, 5], backgroundcolor="rgb(240, 230,240)"),
        zaxis=dict(title='Z Axis', range=[-1, 5], backgroundcolor="rgb(240, 240,230)"),
        camera=dict(eye=dict(x=1.5, y=1.2, z=0.8)) # Great initial angle to see all 3
    ),
    width=950, height=750, margin=dict(l=0, r=0, b=0, t=50)
)
fig.show()
```

## 4. Understanding Rank-1 Matrices

A **nonzero matrix has rank 1** when its row space and column space are both one-dimensional. Equivalently, all its rows are scalar multiples of one nonzero row vector, and all its columns are scalar multiples of one nonzero column vector.

For example,

$$A=\begin{pmatrix}2&12\\3&18\\4&24\end{pmatrix}.$$

**Rows:** Each row is a multiple of $(1,6)$:

$$(2,12)=2(1,6),\quad (3,18)=3(1,6),\quad (4,24)=4(1,6).$$

Thus,

$$\operatorname{Row}(A)=\operatorname{span}\{(1,6)\},$$

a one-dimensional line through the origin in $\mathbb R^2$.

**Columns:** The second column is six times the first:

$$\begin{pmatrix}12\\18\\24\end{pmatrix}
=6\begin{pmatrix}2\\3\\4\end{pmatrix}.$$

Thus,

$$\operatorname{Col}(A)=
\operatorname{span}\left\{\begin{pmatrix}2\\3\\4\end{pmatrix}\right\},$$

a one-dimensional line through the origin in $\mathbb R^3$.

**Important distinction:** The row vectors live in $\mathbb R^2$, while the column vectors live in $\mathbb R^3$. In each space, their span is a line—not a plane. The zero matrix has rank 0, not rank 1.

**Data science connection:** A rank-1 matrix contains only one independent row/column direction. Singular Value Decomposition (SVD) represents a matrix as a sum of rank-1 matrices, which is useful for low-rank approximation and identifying dominant patterns.

## 5. Matrix Multiplication: Four Perspectives

Matrix multiplication $AB = C$ requires compatibility in dimensions: $A_{m \times n}$ and $B_{n \times p}$ yields a result matrix $C_{m \times p}$. There are four rigorous ways to conceptualize and execute this operation.

### Way 1: The Dot Product (Standard) Method
Each entry in the resulting matrix $C$ is computed individually by taking the dot product of a row from $A$ and a column from $B$.

$$ C_{ij} = (\text{row } i \text{ of } A) \cdot (\text{column } j \text{ of } B) $$

Mathematically, this is expressed as a summation:
$$ C_{ij} = \sum_{k=1}^{n} a_{ik}b_{kj}. $$

For instance, $C_{34}$ uses row 3 of $A$ and column 4 of $B$ when those rows and columns exist.

**Detailed Example for Way 1:**
Let $A = \begin{pmatrix} 1 & 2 & 1 \\ 3 & 8 & 1 \\ 0 & 4 & 1 \end{pmatrix}$ and $B = \begin{pmatrix} 1 & 2 \\ 3 & 4 \\ 5 & 6 \end{pmatrix}$. 
To find $C_{21}$ (row 2 of $A$ dotted with col 1 of $B$):
$$ C_{21} = (3)(1) + (8)(3) + (1)(5) = 3 + 24 + 5 = 32 $$

### Way 2: The Column Perspective (Linear Combinations)
Instead of looking at individual entries, this perspective views the matrix $C$ column by column. This uses the **linear combs of column technique** from previous lectures.

The 1st matrix $A$ multiplies the columns of $B$ independently:
$$ \text{1st matrix } A \times (n\text{-th column of 2nd matrix } B) = \text{result matrix } n\text{-th column} $$

Every column in the result matrix $C$ is a linear combination of the columns of $A$, where the scalar coefficients are drawn from the respective column of $B$.

**Detailed Example for Way 2:**
Using the same $A$ and $B$, let's find the entire **first column** of $C$:
$$ C_{\text{col 1}} = A \times B_{\text{col 1}} = \begin{pmatrix} 1 & 2 & 1 \\ 3 & 8 & 1 \\ 0 & 4 & 1 \end{pmatrix} \begin{pmatrix} 1 \\ 3 \\ 5 \end{pmatrix} $$
$$ = 1\begin{pmatrix} 1 \\ 3 \\ 0 \end{pmatrix} + 3\begin{pmatrix} 2 \\ 8 \\ 4 \end{pmatrix} + 5\begin{pmatrix} 1 \\ 1 \\ 1 \end{pmatrix} = \begin{pmatrix} 1 \\ 3 \\ 0 \end{pmatrix} + \begin{pmatrix} 6 \\ 24 \\ 12 \end{pmatrix} + \begin{pmatrix} 5 \\ 5 \\ 5 \end{pmatrix} = \begin{pmatrix} 12 \\ 32 \\ 17 \end{pmatrix} $$

### Way 3: The Row Perspective (Linear Combinations)
Symmetrically, we can view the multiplication row by row. This uses the **linear combs of rows technique**. A single row of $A$ multiplies the entire matrix $B$.

$$ (\text{row } i \text{ of } A) \times B = \text{row } i \text{ of } C $$

Every row in the result matrix $C$ is a linear combination of the rows of $B$, where the scalar weights are provided by the $i$-th row of $A$.

**Detailed Example for Way 3:**
Let's find the **first row** of $C$:
$$ C_{\text{row 1}} = A_{\text{row 1}} \times B = \begin{pmatrix} 1 & 2 & 1 \end{pmatrix} \begin{pmatrix} 1 & 2 \\ 3 & 4 \\ 5 & 6 \end{pmatrix} $$
$$ = 1\begin{pmatrix} 1 & 2 \end{pmatrix} + 2\begin{pmatrix} 3 & 4 \end{pmatrix} + 1\begin{pmatrix} 5 & 6 \end{pmatrix} = \begin{pmatrix} 1 & 2 \end{pmatrix} + \begin{pmatrix} 6 & 8 \end{pmatrix} + \begin{pmatrix} 5 & 6 \end{pmatrix} = \begin{pmatrix} 12 & 16 \end{pmatrix} $$

### Way 4: Columns $\times$ Rows (Sum of Rank-1 Matrices)
The fourth perspective breaks the operation down into a sum of full **Rank-1 Matrices** (as explained in Section 4). 

$$ AB = \text{sum of } (\text{cols of } A) \times (\text{rows of } B) $$

The shape of a column vector $(m \times 1)$ multiplied by a row vector $(1 \times p)$ results in a **full matrix** $(m \times p)$:
$$ (\text{column of } A) \times (\text{row of } B) = \text{a full matrix} $$
$$ (m \times 1) \times (1 \times p) = m \times p $$

Consider the following exact numerical evaluation:
$$ \begin{pmatrix} 2 & 7 \\ 3 & 8 \\ 4 & 9 \end{pmatrix} \begin{pmatrix} 1 & 6 \\ 0 & 0 \end{pmatrix} = \begin{pmatrix} 2 \\ 3 \\ 4 \end{pmatrix} \begin{pmatrix} 1 & 6 \end{pmatrix} + \begin{pmatrix} 7 \\ 8 \\ 9 \end{pmatrix} \begin{pmatrix} 0 & 0 \end{pmatrix} $$

Executing the column-times-row expansion:
$$ = \begin{pmatrix} 2 \cdot 1 & 2 \cdot 6 \\ 3 \cdot 1 & 3 \cdot 6 \\ 4 \cdot 1 & 4 \cdot 6 \end{pmatrix} + \begin{pmatrix} 7 \cdot 0 & 7 \cdot 0 \\ 8 \cdot 0 & 8 \cdot 0 \\ 9 \cdot 0 & 9 \cdot 0 \end{pmatrix} $$

$$ = \begin{pmatrix} 2 & 12 \\ 3 & 18 \\ 4 & 24 \end{pmatrix} + \begin{pmatrix} 0 & 0 \\ 0 & 0 \\ 0 & 0 \end{pmatrix} = \begin{pmatrix} 2 & 12 \\ 3 & 18 \\ 4 & 24 \end{pmatrix} $$

As we can see, multiplying a column of $A$ by a row of $B$ explicitly generated the Rank-1 Matrix $\begin{pmatrix} 2 & 12 \\ 3 & 18 \\ 4 & 24 \end{pmatrix}$ that we analyzed earlier.

## 6. Geometric Interpretation of Way 4 (Rank-1 Matrix)

A single column-times-row product creates a rank-at-most-1 matrix. For the example,

$$\begin{pmatrix}2\\3\\4\end{pmatrix}
\begin{pmatrix}1&6\end{pmatrix}
=
\begin{pmatrix}2&12\\3&18\\4&24\end{pmatrix}.$$

### Row space (in $\mathbb R^2$)

The rows are

$$(2,12)=2(1,6),\quad (3,18)=3(1,6),\quad (4,24)=4(1,6).$$

Therefore,

$$\boxed{\operatorname{Row}(A)=\operatorname{span}\{(1,6)\}}.$$

All row vectors lie on the same line through the origin in $\mathbb R^2$. This is a **1D subspace**, not the whole 2D plane.

### Column space (in $\mathbb R^3$)

The columns are

$$\begin{pmatrix}2\\3\\4\end{pmatrix},
\qquad
\begin{pmatrix}12\\18\\24\end{pmatrix}
=6\begin{pmatrix}2\\3\\4\end{pmatrix}.$$

Therefore,

$$\boxed{\operatorname{Col}(A)=
\operatorname{span}\left\{\begin{pmatrix}2\\3\\4\end{pmatrix}\right\}}.$$

Both columns lie on the same line through the origin in $\mathbb R^3$. They do **not** span a plane, because the second column is not an independent direction.

**Key idea:** The ambient space tells us where the vectors live ($\mathbb R^2$ for rows and $\mathbb R^3$ for columns); the span tells us the subspace they actually generate. Here both spaces have dimension 1.

## 7. Visualizations: 2D Row Space and 3D Column Space

The first plot shows the row space as a line in $\mathbb R^2$. The second shows the column space as a line in $\mathbb R^3$—no plane is implied.

```python
import numpy as np
import matplotlib.pyplot as plt

# The rows are scalar multiples of the base row vector (1, 6).
base_vector = np.array([1, 6])
multipliers = [2, 3, 4]
vectors = [base_vector * m for m in multipliers]

fig, ax = plt.subplots(figsize=(8, 8))
ax.axhline(0, color="black", linewidth=1)
ax.axvline(0, color="black", linewidth=1)

# Draw the full 1D row-space line through the origin.
t = np.linspace(-1, 4.5, 200)
ax.plot(t * base_vector[0], t * base_vector[1],
        color="gray", linestyle="--", label="Row space: span{(1, 6)}")

colors = ["blue", "green", "purple"]
for v, color in zip(vectors, colors):
    ax.quiver(0, 0, v[0], v[1], angles="xy", scale_units="xy", scale=1,
              color=color, width=0.006, label=f"({v[0]}, {v[1]})")

ax.set_xlim(-5, 10)
ax.set_ylim(-5, 55)
ax.set_aspect("equal", adjustable="box")
ax.grid(True, linestyle="--", alpha=0.5)
ax.set_title("Row Space: a 1D line in R²")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
plt.show()
```

```python
import plotly.graph_objects as go
import numpy as np

# The second column is exactly 6 times the first.
col1 = np.array([2, 3, 4])
col2 = 6 * col1

fig = go.Figure()

# Draw the column-space line through the origin.
t = np.linspace(-1.2, 7.5, 100)
line_points = t[:, None] * col1[None, :]
fig.add_trace(go.Scatter3d(
    x=line_points[:, 0], y=line_points[:, 1], z=line_points[:, 2],
    mode="lines",
    line=dict(color="gray", width=4, dash="dash"),
    name="Column space: span{(2, 3, 4)}"
))

# First column
fig.add_trace(go.Scatter3d(
    x=[0, col1[0]], y=[0, col1[1]], z=[0, col1[2]],
    mode="lines+markers+text",
    line=dict(color="blue", width=7),
    marker=dict(size=4),
    text=["", "(2, 3, 4)"],
    textposition="top center",
    name="Column 1"
))

# Second column: same direction, six times as long
fig.add_trace(go.Scatter3d(
    x=[0, col2[0]], y=[0, col2[1]], z=[0, col2[2]],
    mode="lines+markers+text",
    line=dict(color="red", width=5, dash="dash"),
    marker=dict(size=4),
    text=["", "(12, 18, 24)"],
    textposition="top center",
    name="Column 2 = 6 × Column 1"
))

fig.update_layout(
    title="Column Space: a 1D line through the origin in R³",
    scene=dict(
        xaxis=dict(title="X"),
        yaxis=dict(title="Y"),
        zaxis=dict(title="Z"),
        aspectmode="data",
        camera=dict(eye=dict(x=1.6, y=1.6, z=1.3))
    ),
    width=900,
    height=700,
    margin=dict(l=0, r=0, b=0, t=50)
)
fig.show()
```

## 8. Block Matrix Multiplication

Matrices can be partitioned into smaller matrices, or **blocks**, and multiplied according to the usual matrix multiplication rules. The block dimensions must be compatible.

Suppose

$$A=\begin{pmatrix}A_1&A_2\\A_3&A_4\end{pmatrix},
\qquad
B=\begin{pmatrix}B_1&B_2\\B_3&B_4\end{pmatrix}.$$

For this block product to be valid, the block sizes must align so that each product below is defined and the two terms being added in each entry have the same dimensions. For example, in the conformable partition,

- $A_1B_1$ and $A_2B_3$ must both be defined and have the same shape;
- similarly for the other three block entries.

Then

$$AB=
\begin{pmatrix}
A_1B_1+A_2B_3&A_1B_2+A_2B_4\\
A_3B_1+A_4B_3&A_3B_2+A_4B_4
\end{pmatrix}.$$

This is structurally the same row-by-column rule as ordinary matrix multiplication, except that the entries are themselves matrices. Identical partition boundaries are a convenient sufficient setup for square matrices, but the essential requirement is **compatible block dimensions**, not merely that both matrices are square.

## 9. Practical Application: Which Method to Use When?

Understanding four different ways to compute $AB$ provides necessary tools for distinct mathematical contexts:

- **Way 1 (Dot Product):** Implement this method when writing computer algorithms for standard matrix multiplication or when manually calculating a specific, isolated entry (e.g., finding only $C_{34}$ without computing the rest of the matrix).
- **Way 2 (Column Perspective):** Implement this when solving systems of equations $A\mathbf{x} = \mathbf{b}$. It proves that multiplying a matrix by a vector is fundamentally taking a linear combination of the columns of that matrix.
- **Way 3 (Row Perspective):** Implement this when analyzing row operations, such as identifying the row space or performing elimination steps (where matrices act on the left side to combine rows).
- **Way 4 (Sum of Rank-1 Matrices):** Implement this perspective in advanced data science, signal processing, and matrix factorizations. Disassembling a matrix into sums of column-times-row Rank-1 matrices is the foundational mechanic behind the Singular Value Decomposition (SVD).
