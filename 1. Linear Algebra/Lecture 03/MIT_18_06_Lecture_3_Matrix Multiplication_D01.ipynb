## 1. Geometric Meaning of the Dot Product

Before evaluating matrix multiplication, the fundamental operation governing these interactions—the **dot product**—must be understood geometrically. 

For two vectors $\mathbf{A}$ and $\mathbf{B}$, the dot product is defined geometrically as:
$$ \mathbf{A} \cdot \mathbf{B} = \|\mathbf{A}\| \|\mathbf{B}\| \cos(\theta) $$
where $\theta$ is the angle between them. 

Geometrically, the dot product represents the **projection** of one vector onto another. However, **a dot product is always a scalar (a single number), not a vector.** It acts as an "Interaction Score" indicating how much of vector $\mathbf{A}$ aligns with vector $\mathbf{B}$.

**The Torchlight and Shadow Analogy:**
Imagine vector $\mathbf{B}$ is the flat ground, and vector $\mathbf{A}$ is a pole sticking out of the origin (0,0) at some angle. If you hold a torchlight exactly above the tip of vector $\mathbf{A}$ and shine it straight down (perpendicularly) toward the ground ($\mathbf{B}$), the resulting shadow is the projection.

The value of the scalar dot product dictates exactly how this shadow behaves:

1.  **Positive Dot Product (+):** The angle $\theta$ is acute ($< 90^\circ$). The pole leans forward. The torchlight casts a shadow in the *same* (positive) direction as vector $\mathbf{B}$.
2.  **Zero Dot Product (0):** The angle $\theta$ is exactly $90^\circ$ (orthogonal/perpendicular). The pole stands straight up. The torchlight shines straight down to the origin point (0,0). **No shadow is cast along $\mathbf{B}$.** The vectors have absolutely no influence on each other.
3.  **Negative Dot Product (-):** The angle $\theta$ is obtuse ($> 90^\circ$). The pole leans backward. The torchlight casts a shadow behind the origin, in the *opposite* (negative) direction of vector $\mathbf{B}$.

## 2. Scalar Component vs. Vector Projection

It is crucial to distinguish between the **length of the shadow** and the **actual physical shadow (the vector)**.

**1. The Scalar Component (Length of the Shadow):**
If we want to know *how much* of vector $\mathbf{A}$ goes in the direction of vector $\mathbf{B}$, we are looking for its scalar component. From basic trigonometry, this length is $\|\mathbf{A}\| \cos(\theta)$. 
Using the dot product formula, we can isolate this length:
$$ \|\mathbf{A}\| \cos(\theta) = \frac{\mathbf{A} \cdot \mathbf{B}}{\|\mathbf{B}\|} $$
This is just a number (e.g., 3 units long). It has no direction.

**2. The Vector Projection (The Actual Shadow):**
A physical shadow pointing along the ground is a **vector**. To convert our scalar length into a proper vector, we must multiply it by the *direction* of the ground. 
The direction of the ground is given by the unit vector of $\mathbf{B}$, which is $\frac{\mathbf{B}}{\|\mathbf{B}\|}$.

Multiplying the length by the direction gives us the final Vector Projection formula:
$$ \text{Vector Projection} = (\text{Scalar Length}) \times (\text{Unit Vector Direction}) $$
$$ \text{Vector Projection} = \left( \frac{\mathbf{A} \cdot \mathbf{B}}{\|\mathbf{B}\|} \right) \times \left( \frac{\mathbf{B}}{\|\mathbf{B}\|} \right) $$
$$ \text{Vector Projection} = \frac{\mathbf{A} \cdot \mathbf{B}}{\|\mathbf{B}\|^2} \mathbf{B} $$

This formula perfectly explains how the scalar dot product ($\mathbf{A} \cdot \mathbf{B}$) acts as the engine that scales the final projection vector.

## 4. Understanding Rank-1 Matrices

Before diving into "Way 4" of matrix multiplication, we must understand the concept of a Rank-1 Matrix. 

A **Rank-1 Matrix** is a highly structured matrix where **every row is just a multiple of the exact same base row vector** (and consequently, every column is just a multiple of the same base column vector).

For example, consider this $3 \times 2$ matrix:
$$ \begin{pmatrix} 2 & 12 \\ 3 & 18 \\ 4 & 24 \end{pmatrix} $$

- **Looking at the rows:** The first row is $(2, 12)$. The second row $(3, 18)$ is just $1.5 \times (2, 12)$. The third row $(4, 24)$ is exactly $2 \times (2, 12)$. All rows are generated from one fundamental piece of information.
- **Looking at the columns:** The first column is $\begin{pmatrix} 2 \\ 3 \\ 4 \end{pmatrix}$. The second column is $\begin{pmatrix} 12 \\ 18 \\ 24 \end{pmatrix}$, which is precisely $6 \times \begin{pmatrix} 2 \\ 3 \\ 4 \end{pmatrix}$.

**Geometric Implication:**
Because every vector in a Rank-1 Matrix is a scaled version of one base vector, they do not spread out to form planes or 3D spaces. If you plot all the row vectors, they will all lie exactly on a single 1D line. 

**Data Science Implication:**
A Rank-1 Matrix might look large and full of numbers, but its actual "information content" is extremely low. It contains only one independent direction of data. In algorithms like SVD (Singular Value Decomposition), huge datasets are broken down into sums of Rank-1 matrices to isolate the most important patterns.

## 5. Matrix Multiplication: Four Perspectives

Matrix multiplication $AB = C$ requires compatibility in dimensions: $A_{m \times n}$ and $B_{n \times p}$ yields a result matrix $C_{m \times p}$. There are four rigorous ways to conceptualize and execute this operation.

### Way 1: The Dot Product (Standard) Method
Each entry in the resulting matrix $C$ is computed individually by taking the dot product of a row from $A$ and a column from $B$.

$$ C_{34} = (\text{row } 3 \text{ of } A) \cdot (\text{col } 4 \text{ of } B) $$

Mathematically, this is expressed as a summation:
$$ C_{34} = a_{31} \cdot b_{14} + a_{32} \cdot b_{24} + a_{33} \cdot b_{34} + \dots = \sum_{k=1}^{n} a_{3k}b_{k4} $$

**Detailed Example for Way 1:**
Let $A = \begin{pmatrix} 1 & 2 & 1 \\ 3 & 8 & 1 \\ 0 & 4 & 1 \end{pmatrix}$ and $B = \begin{pmatrix} 1 & 2 \\ 3 & 4 \\ 5 & 6 \end{pmatrix}$. 
To find $C_{21}$ (row 2 of $A$ dotted with col 1 of $B$):
$$ C_{21} = (3)(1) + (8)(3) + (1)(5) = 3 + 24 + 5 = 32 $$

### Way 2: The Column Perspective (Linear Combinations)
Instead of looking at individual entries, this perspective views the matrix $C$ column by column. This uses the **linear combs of column technique** from previous lectures.

The 1st matrix $A$ multiplies the columns of $B$ independently:
$$ \text{1st matrix } A \times (n\text{-th column of 2nd matrix } B) = \text{result matrix } n\text{-th column} $$

Every column in the result matrix $C$ is a linear combination of the columns of $A$, where the scalar coefficients are drawn from the respective column of $B$.

**Detailed Example for Way 2:**
Using the same $A$ and $B$, let's find the entire **first column** of $C$:
$$ C_{\text{col 1}} = A \times B_{\text{col 1}} = \begin{pmatrix} 1 & 2 & 1 \\ 3 & 8 & 1 \\ 0 & 4 & 1 \end{pmatrix} \begin{pmatrix} 1 \\ 3 \\ 5 \end{pmatrix} $$
$$ = 1\begin{pmatrix} 1 \\ 3 \\ 0 \end{pmatrix} + 3\begin{pmatrix} 2 \\ 8 \\ 4 \end{pmatrix} + 5\begin{pmatrix} 1 \\ 1 \\ 1 \end{pmatrix} = \begin{pmatrix} 1 \\ 3 \\ 0 \end{pmatrix} + \begin{pmatrix} 6 \\ 24 \\ 12 \end{pmatrix} + \begin{pmatrix} 5 \\ 5 \\ 5 \end{pmatrix} = \begin{pmatrix} 12 \\ 32 \\ 17 \end{pmatrix} $$

### Way 3: The Row Perspective (Linear Combinations)
Symmetrically, we can view the multiplication row by row. This uses the **linear combs of rows technique**. A single row of $A$ multiplies the entire matrix $B$.

$$ (\text{row } i \text{ of } A) \times B = \text{row } i \text{ of } C $$

Every row in the result matrix $C$ is a linear combination of the rows of $B$, where the scalar weights are provided by the $i$-th row of $A$.

**Detailed Example for Way 3:**
Let's find the **first row** of $C$:
$$ C_{\text{row 1}} = A_{\text{row 1}} \times B = \begin{pmatrix} 1 & 2 & 1 \end{pmatrix} \begin{pmatrix} 1 & 2 \\ 3 & 4 \\ 5 & 6 \end{pmatrix} $$
$$ = 1\begin{pmatrix} 1 & 2 \end{pmatrix} + 2\begin{pmatrix} 3 & 4 \end{pmatrix} + 1\begin{pmatrix} 5 & 6 \end{pmatrix} = \begin{pmatrix} 1 & 2 \end{pmatrix} + \begin{pmatrix} 6 & 8 \end{pmatrix} + \begin{pmatrix} 5 & 6 \end{pmatrix} = \begin{pmatrix} 12 & 16 \end{pmatrix} $$

### Way 4: Columns $\times$ Rows (Sum of Rank-1 Matrices)
The fourth perspective breaks the operation down into a sum of full **Rank-1 Matrices** (as explained in Section 4). 

$$ AB = \text{sum of } (\text{cols of } A) \times (\text{rows of } B) $$

The shape of a column vector $(m \times 1)$ multiplied by a row vector $(1 \times p)$ results in a **full matrix** $(m \times p)$:
$$ (\text{column of } A) \times (\text{row of } B) = \text{a full matrix} $$
$$ (m \times 1) \times (1 \times p) = m \times p $$

Consider the following exact numerical evaluation:
$$ \begin{pmatrix} 2 & 7 \\ 3 & 8 \\ 4 & 9 \end{pmatrix} \begin{pmatrix} 1 & 6 \\ 0 & 0 \end{pmatrix} = \begin{pmatrix} 2 \\ 3 \\ 4 \end{pmatrix} \begin{pmatrix} 1 & 6 \end{pmatrix} + \begin{pmatrix} 7 \\ 8 \\ 9 \end{pmatrix} \begin{pmatrix} 0 & 0 \end{pmatrix} $$

Executing the column-times-row expansion:
$$ = \begin{pmatrix} 2 \cdot 1 & 2 \cdot 6 \\ 3 \cdot 1 & 3 \cdot 6 \\ 4 \cdot 1 & 4 \cdot 6 \end{pmatrix} + \begin{pmatrix} 7 \cdot 0 & 7 \cdot 0 \\ 8 \cdot 0 & 8 \cdot 0 \\ 9 \cdot 0 & 9 \cdot 0 \end{pmatrix} $$

$$ = \begin{pmatrix} 2 & 12 \\ 3 & 18 \\ 4 & 24 \end{pmatrix} + \begin{pmatrix} 0 & 0 \\ 0 & 0 \\ 0 & 0 \end{pmatrix} = \begin{pmatrix} 2 & 12 \\ 3 & 18 \\ 4 & 24 \end{pmatrix} $$

As we can see, multiplying a column of $A$ by a row of $B$ explicitly generated the Rank-1 Matrix $\begin{pmatrix} 2 & 12 \\ 3 & 18 \\ 4 & 24 \end{pmatrix}$ that we analyzed earlier.

## 6. Geometric Interpretation of Way 4 (Rank-1 Matrix)

The result of a single column-times-row multiplication produces a highly structured Rank-1 Matrix where columns and rows align along 1D lines.

**Row Space Interpretation (2D Plane):**
$$ \text{Now } \begin{pmatrix} 2 \\ 3 \\ 4 \end{pmatrix}_{m \times 1} \begin{pmatrix} 1 & 6 \end{pmatrix}_{1 \times p} = \begin{pmatrix} 2 & 12 \\ 3 & 18 \\ 4 & 24 \end{pmatrix}_{\\text{result matrix } m \times p} $$

In that $m \times p$ result matrix, each row is a multiple of $2(1, 6), 3(1, 6), 4(1, 6)$.
So if we draw a row vector $(1, 6)$ and:
- $2(1, 6) \Rightarrow (2, 12)$
- $3(1, 6) \Rightarrow (3, 18)$
- $4(1, 6) \Rightarrow (4, 24)$

In a 2D plane, then $(3, 18), (2, 12), (4, 24)$ will go through $(1, 6)$ or they will lie on the same line and plane because $(2, 12), (3, 18), (4, 24)$ all those are multiple of $(1, 6)$.

**Column Space Interpretation (3D Plane):**
In that $m \times p$ result matrix, each col is a multiple of:
- $1 \cdot \begin{pmatrix} 2 \\ 3 \\ 4 \end{pmatrix} \Rightarrow \begin{pmatrix} 2 \\ 3 \\ 4 \end{pmatrix}$
- $6 \cdot \begin{pmatrix} 2 \\ 3 \\ 4 \end{pmatrix} \Rightarrow \begin{pmatrix} 12 \\ 18 \\ 24 \end{pmatrix}$

So if we draw a col vector $\begin{pmatrix} 2 \\ 3 \\ 4 \end{pmatrix}$ and $\begin{pmatrix} 12 \\ 18 \\ 24 \end{pmatrix}$, then in a 3D plane $\begin{pmatrix} 2 \\ 3 \\ 4 \end{pmatrix}$ and $\begin{pmatrix} 12 \\ 18 \\ 24 \end{pmatrix}$ will lie on the same line and plane because $\begin{pmatrix} 12 \\ 18 \\ 24 \end{pmatrix}$ is a multiple of $\begin{pmatrix} 2 \\ 3 \\ 4 \end{pmatrix}$.

## 7. Visualizations (2D Row Space & 3D Column Space)

```python
import numpy as np
import matplotlib.pyplot as plt

# Define row vectors
base_vector = np.array([1, 6])
multipliers = [2, 3, 4]
vectors = [base_vector * m for m in multipliers]

# Create 2D plot for Row Space
plt.figure(figsize=(8, 8))
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

# Plot origin to base vector
plt.quiver(0, 0, base_vector[0], base_vector[1], angles='xy', scale_units='xy', scale=1, color='black', width=0.005, label='(1, 6)')

# Plot origin to multiple vectors (matching the handwritten note's line graph)
colors = ['blue', 'green', 'purple']
for i, (v, color) in enumerate(zip(vectors, colors)):
    plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color=color, alpha=0.6, width=0.004, label=f'({v[0]}, {v[1]})')
    plt.plot([0, v[0]], [0, v[1]], color='gray', linestyle='--', alpha=0.5)

# Annotations & Formatting
plt.xlim(-1, 5)
plt.ylim(-2, 26)
plt.grid(True, linestyle='--', alpha=0.7)
plt.title("Row Space Interpretation (2D Plane)\nAll rows are multiples of (1, 6) and lie on the same 1D line", pad=20)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

```python
import plotly.graph_objects as go
import numpy as np

# Define column vectors
col1 = np.array([2, 3, 4])
col2 = np.array([12, 18, 24]) # 6 * col1

fig = go.Figure()

# Add trace for col1
fig.add_trace(go.Scatter3d(
    x=[0, col1[0]], y=[0, col1[1]], z=[0, col1[2]],
    mode='lines+markers+text',
    line=dict(color='blue', width=6),
    marker=dict(size=4),
    text=["", "(2, 3, 4)"],
    textposition="top center",
    name="1 * (2, 3, 4)"
))

# Add trace for col2
fig.add_trace(go.Scatter3d(
    x=[0, col2[0]], y=[0, col2[1]], z=[0, col2[2]],
    mode='lines+markers+text',
    line=dict(color='red', width=3, dash='dash'),
    marker=dict(size=6),
    text=["", "(12, 18, 24)"],
    textposition="top center",
    name="6 * (2, 3, 4)"
))

# Add the 3D plane that both vectors lie on
# We create a meshgrid for the plane based on a scaling factor for the vector
s_vals = np.linspace(-1, 8, 20)
# Since the vectors are colinear, the "plane" is technically just a line in 3D space.
# To show a plane, we can cross it with the z-axis (or another vector) to give it visual width.
v1 = np.array([2, 3, 4])
v2 = np.array([0, 1, 0]) # Arbitrary vector to define a plane containing the line
normal = np.cross(v1, v2)

# Create a grid to plot the plane
xx, yy = np.meshgrid(np.linspace(-2, 15, 10), np.linspace(-2, 20, 10))
# Equation of plane: normal[0]*x + normal[1]*y + normal[2]*z = 0
# z = (-normal[0]*x - normal[1]*y) / normal[2]
zz = (-normal[0]*xx - normal[1]*yy) / normal[2]

fig.add_trace(go.Surface(
    x=xx, y=yy, z=zz,
    opacity=0.3,
    colorscale='Greys',
    showscale=False,
    name='Span Plane'
))

fig.update_layout(
    title='Column Space Interpretation (360° 3D Plane)<br>Both vectors lie on the exact same 1D line and span the same plane',
    scene=dict(
        xaxis=dict(title='X Axis', backgroundcolor="rgb(200, 200, 230)"),
        yaxis=dict(title='Y Axis', backgroundcolor="rgb(230, 200,230)"),
        zaxis=dict(title='Z Axis', backgroundcolor="rgb(230, 230,200)"),
        camera=dict(
            eye=dict(x=1.5, y=1.5, z=1.5) # Initial 3D view angle
        )
    ),
    width=900,
    height=700,
    margin=dict(l=0, r=0, b=0, t=50)
)
fig.show()
```

## 8. Block Matrix Multiplication

Matrices can be partitioned into smaller matrices, or **blocks**, and multiplied according to the same fundamental rules as standard scalar multiplication, provided the dimensions of the blocks are compatible.

Suppose square matrix $A$ and square matrix $B$ are partitioned identically:
$$ A = \begin{pmatrix} A_1 & A_2 \\ A_3 & A_4 \end{pmatrix}, \quad B = \begin{pmatrix} B_1 & B_2 \\ B_3 & B_4 \end{pmatrix} $$

The product $AB$ can be found by treating the blocks as individual elements (row times column block multiplication):
$$ AB = \begin{pmatrix} A_1B_1 + A_2B_3 & A_1B_2 + A_2B_4 \\ A_3B_1 + A_4B_3 & A_3B_2 + A_4B_4 \end{pmatrix} $$

This is structurally identical to standard dot product matrix multiplication (Way 1) but operates on matrices instead of scalars.

## 9. Practical Application: Which Method to Use When?

Understanding four different ways to compute $AB$ provides necessary tools for distinct mathematical contexts:

- **Way 1 (Dot Product):** Implement this method when writing computer algorithms for standard matrix multiplication or when manually calculating a specific, isolated entry (e.g., finding only $C_{34}$ without computing the rest of the matrix).
- **Way 2 (Column Perspective):** Implement this when solving systems of equations $A\mathbf{x} = \mathbf{b}$. It proves that multiplying a matrix by a vector is fundamentally taking a linear combination of the columns of that matrix.
- **Way 3 (Row Perspective):** Implement this when analyzing row operations, such as identifying the row space or performing elimination steps (where matrices act on the left side to combine rows).
- **Way 4 (Sum of Rank-1 Matrices):** Implement this perspective in advanced data science, signal processing, and matrix factorizations. Disassembling a matrix into sums of column-times-row Rank-1 matrices is the foundational mechanic behind the Singular Value Decomposition (SVD).
