## 1. The Fundamental Problem of Linear Algebra

The central and most fundamental problem of linear algebra is to solve a system of linear equations:

$$\text{Solve } n \text{ linear equations with } n \text{ unknowns. } (Ax = b)$$

Prof. Gilbert Strang introduces three distinct ways to view and interpret this system:
1. **The Row Picture:** Looking at the problem one equation (row) at a time (e.g., intersecting lines in 2D or flat planes in 3D).
2. **The Column Picture:** Looking at the problem a column at a time as a **linear combination of column vectors** (the central viewpoint of the entire course).
3. **The Matrix Form ($Ax = b$):** Expressing the system cleanly in algebraic form using a coefficient matrix $A$, unknown vector $x$, and right-hand side vector $b$.

## 2. A 2-Equation, 2-Unknown ($2 \times 2$) Example

Consider the system of two linear equations in two unknowns ($x$ and $y$):

$$\begin{aligned}
2x - y &= 0 \quad &\text{--- (Equation 1)} \\
-x + 2y &= 3 \quad &\text{--- (Equation 2)}
\end{aligned}$$

### Matrix Form: $Ax = b$
Writing this system in matrix-vector notation:

$$\begin{bmatrix} 2 & -1 \\ -1 & 2 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 0 \\ 3 \end{bmatrix}$$

Where:
- **Coefficient Matrix ($A$):** $A = \begin{bmatrix} 2 & -1 \\ -1 & 2 \end{bmatrix}$ (contains the coefficients of $x$ and $y$).
- **Vector of Unknowns ($x$):** $\mathbf{x} = \begin{bmatrix} x \\ y \end{bmatrix}$.
- **Right-Hand Side Vector ($b$):** $\mathbf{b} = \begin{bmatrix} 0 \\ 3 \end{bmatrix}$.

## 3. The Row Picture in 2D

In the **Row Picture**, we plot each linear equation as a separate line in the 2D Cartesian plane ($\mathbb{R}^2$). The solution is the single point $(x, y)$ that lies on **both** lines simultaneously (the intersection).

### Equation (i): $2x - y = 0 \iff y = 2x$
- Passes directly through the origin $(0, 0)$ because the right-hand constant is $0$.
- Key points on Line 1:
  - At $x = 0 \implies y = 0 \implies (0, 0)$
  - At $x = 1 \implies y = 2 \implies (1, 2)$
  - At $x = \frac{1}{2} \implies y = 1 \implies (0.5, 1)$
  - Slope $m_1 = 2$.

### Equation (ii): $-x + 2y = 3 \iff y = \frac{1}{2}x + \frac{3}{2}$
- Does not pass through origin because $(0, 0)$ gives $0 \neq 3$.
- Key points on Line 2:
  - When $y = 0 \implies -x = 3 \implies x = -3 \implies (-3, 0)$
  - When $x = -1 \implies -(-1) + 2y = 3 \implies 2y = 2 \implies y = 1 \implies (-1, 1)$
  - When $x = 1 \implies -(1) + 2y = 3 \implies 2y = 4 \implies y = 2 \implies (1, 2)$
  - Slope $m_2 = \frac{1}{2}$.

### Intersection & Verification
Both lines intersect at exactly one common point:
$$\mathbf{(x, y) = (1, 2)}$$

- Eq 1: $2(1) - (2) = 0$ (Satisfied)
- Eq 2: $-(1) + 2(2) = 3$ (Satisfied)

```python
# 4. Exact Visualization: 2D Row Picture
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 3, 400)

# Line 1: 2x - y = 0 => y = 2x
y1 = 2 * x

# Line 2: -x + 2y = 3 => y = 0.5*x + 1.5
y2 = 0.5 * x + 1.5

# Increased both width and height
plt.figure(figsize=(14, 10))
plt.plot(x, y1, label=r'Line 1: $2x - y = 0$', color='blue', linewidth=2)
plt.plot(x, y2, label=r'Line 2: $-x + 2y = 3$', color='red', linewidth=2)

# Highlight intersection point (1, 2)
plt.plot(1, 2, 'go', markersize=9, label='Intersection / Solution (1, 2)')
plt.annotate('(1, 2)', xy=(1, 2), xytext=(1.2, 2.1), fontsize=11, fontweight='bold', color='green')

# Highlight points from handwritten notes
plt.plot(0, 0, 'bo', markersize=6)
plt.annotate('(0, 0)', xy=(0, 0), xytext=(0.1, -0.4), fontsize=10, color='blue')

plt.plot(-3, 0, 'ro', markersize=6)
plt.annotate('(-3, 0)', xy=(-3, 0), xytext=(-3.5, 0.3), fontsize=10, color='red')

plt.plot(-1, 1, 'ro', markersize=6)
plt.annotate('(-1, 1)', xy=(-1, 1), xytext=(-1.8, 1.2), fontsize=10, color='red')

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.grid(True, linestyle='--', alpha=0.6)

# Keep unit boxes accurate and equal in aspect ratio
plt.gca().set_aspect('equal', adjustable='box')

plt.xlim(-4, 3)
plt.ylim(-2, 5)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Row Picture: Intersection of Two Lines at (1, 2)', fontsize=14, pad=12)
plt.legend(loc='upper left', fontsize=10)
plt.show()
```

## 5. The Column Picture (The Heart of Linear Algebra)

While the row picture looks at intersecting lines or planes, the **Column Picture** looks at the matrix as a set of separate column vectors.

### The Problem Restated in Columns
Consider the coefficient matrix $A$:
$$A = \begin{bmatrix} 2 & -1 \\ -1 & 2 \end{bmatrix}$$

Split matrix $A$ into its two column vectors:
- First column vector: $\mathbf{c}_1 = \begin{bmatrix} 2 \\ -1 \end{bmatrix}$
- Second column vector: $\mathbf{c}_2 = \begin{bmatrix} -1 \\ 2 \end{bmatrix}$
- Target vector (RHS): $\mathbf{b} = \begin{bmatrix} 0 \\ 3 \end{bmatrix}$
- Unknown scalars: $x$ and $y$

### Linear Combination of Columns
The fundamental operation in linear algebra is forming a **linear combination** of vectors:
$$x \mathbf{c}_1 + y \mathbf{c}_2 = \mathbf{b}$$

Writing this out component-by-component:
$$x \begin{bmatrix} 2 \\ -1 \end{bmatrix} + y \begin{bmatrix} -1 \\ 2 \end{bmatrix} = \begin{bmatrix} 0 \\ 3 \end{bmatrix} \quad \text{--- (Equation i)}$$

**Goal:** Find scalar values $x$ and $y$ (scaling factors) such that scaling $\mathbf{c}_1$ by $x$, scaling $\mathbf{c}_2$ by $y$, and adding them together yields the target vector $\mathbf{b}$.

## 6. Scaling and Vector Addition

### 1. Nature of the Terms
* **Vectors in 2D Space ($\mathbb{R}^2$):** $\mathbf{c}_1$ and $\mathbf{c}_2$ represent our base directions.
* **Real Scalars:** $x$ and $y$ act as the scaling factors (coefficients).
* **Scaled Vectors ($x\mathbf{c}_1$ and $y\mathbf{c}_2$):** These represent the original vectors stretched, compressed, or reversed in direction.
* **Vector Sum ($x\mathbf{c}_1 + y\mathbf{c}_2$):** The result of adding these scaled vectors together to reach a new destination in the plane.

### 2. Applying the Known Solution ($x = 1, y = 2$)
* **Scaled First Column:**
  $$\mathbf{c}_1' = x \mathbf{c}_1 = 1 \begin{bmatrix} 2 \\ -1 \end{bmatrix} = \begin{bmatrix} 2 \\ -1 \end{bmatrix}$$
* **Scaled Second Column:**
  $$\mathbf{c}_2' = y \mathbf{c}_2 = 2 \begin{bmatrix} -1 \\ 2 \end{bmatrix} = \begin{bmatrix} -2 \\ 4 \end{bmatrix}$$

### 3. Vector Addition (Head-to-Tail Method)
$$\mathbf{c}_1' + \mathbf{c}_2' = \begin{bmatrix} 2 \\ -1 \end{bmatrix} + \begin{bmatrix} -2 \\ 4 \end{bmatrix} = \begin{bmatrix} 2 + (-2) \\ -1 + 4 \end{bmatrix} = \begin{bmatrix} 0 \\ 3 \end{bmatrix} = \mathbf{b}$$

> **Result:** The linear combination $1\mathbf{c}_1 + 2\mathbf{c}_2$ lands precisely on the target vector $\mathbf{b} = \begin{bmatrix} 0 \\ 3 \end{bmatrix}$.

```python
# 7. Exact Visualization: 2D Column Picture (Corrected LaTeX Warning)
import numpy as np
import matplotlib.pyplot as plt

c1 = np.array([2, -1])
c2 = np.array([-1, 2])
b = np.array([0, 3])

# Scaled vectors
c1_prime = 1 * c1
c2_prime = 2 * c2

plt.figure(figsize=(14, 10))

# Plot c1_prime starting from the origin
plt.quiver(0, 0, c1_prime[0], c1_prime[1], angles='xy', scale_units='xy', scale=1,
           color='blue', width=0.006, label='$c_1\\prime = 1c_1 = [2, -1]^T$')

# Head-to-tail addition: Plot c2_prime starting from the tip of c1_prime
plt.quiver(c1_prime[0], c1_prime[1], c2_prime[0], c2_prime[1], angles='xy', scale_units='xy', scale=1,
           color='orange', width=0.006, label='$c_2\\prime = 2c_2 = [-2, 4]^T$ (Shifted)')

# Plot resulting target vector b from the origin
plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1,
           color='green', width=0.008, label='Target $b = [0, 3]^T$')

# Point markers and annotations
plt.plot(c1_prime[0], c1_prime[1], 'bo')
plt.annotate('$c_1\\prime (2, -1)$', xy=(c1_prime[0], c1_prime[1]), xytext=(2.1, -1.2), fontsize=11, color='blue')

plt.plot(b[0], b[1], 'go')
plt.annotate('$b (0, 3)$', xy=(b[0], b[1]), xytext=(0.1, 3.1), fontsize=12, fontweight='bold', color='green')

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.grid(True, linestyle='--', alpha=0.6)

# Ensure 1:1 aspect ratio
plt.gca().set_aspect('equal', adjustable='box')

plt.xlim(-3, 4)
plt.ylim(-2, 5)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Column Picture: Linear Combination $c_1\\prime + c_2\\prime = b$', fontsize=14, pad=12)
plt.legend(loc='lower right', fontsize=11)
plt.show()
```

## 8. Summary of Vector Addition on 2D Plane

From the previous column picture:
- Shifting scaled $\mathbf{c}_2' = 2\mathbf{c}_2 = \begin{bmatrix} -2 \\ 4 \end{bmatrix}$ to the tip of $\mathbf{c}_1' = \begin{bmatrix} 2 \\ -1 \end{bmatrix}$ using head-to-tail vector addition lands precisely on $(0, 3)$, which equals the right-hand side vector $\mathbf{b}$.
- **Core Takeaway:** In a 2D plane ($\mathbb{R}^2$), plotting all points $(x, y)$ that satisfy one linear equation produces a straight line.

## 9. 3 Linear Equations and 3 Unknowns ($3 \times 3$ System)

Prof. Strang now scales up to three equations and three unknowns ($x, y, z$ in 3-dimensional space $\mathbb{R}^3$):

$$\begin{aligned}
2x - y &= 0 \quad &\text{--- (Equation i)} \\
-x + 2y - z &= -1 \quad &\text{--- (Equation ii)} \\
-3y + 4z &= 4 \quad &\text{--- (Equation iii)}
\end{aligned}$$

### Matrix Form: $Ax = b$
Writing the full system in matrix notation:

$$\begin{bmatrix} 2 & -1 & 0 \\ -1 & 2 & -1 \\ 0 & -3 & 4 \end{bmatrix} \begin{bmatrix} x \\ y \\ z \end{bmatrix} = \begin{bmatrix} 0 \\ -1 \\ 4 \end{bmatrix}$$

Where:
- **Coefficient Matrix ($A$):**
  $$A = \begin{bmatrix} 2 & -1 & 0 \\ -1 & 2 & -1 \\ 0 & -3 & 4 \end{bmatrix}$$
- **Vector of Unknowns ($x$):**
  $$\mathbf{x} = \begin{bmatrix} x \\ y \\ z \end{bmatrix}$$
- **Right-Hand Side Vector ($b$):**
  $$\mathbf{b} = \begin{bmatrix} 0 \\ -1 \\ 4 \end{bmatrix}$$

## 10. The Row Picture in 3 Dimensions ($\mathbb{R}^3$)

In 3D space, every single linear equation represents a **flat 2-dimensional plane**:
- **Equation (i):** $2x - y + 0z = 0$ is a plane passing through the origin $(0, 0, 0)$ (since $z$ is free to be anything).
- **Equation (ii):** $-x + 2y - z = -1$ is a second plane (not passing through the origin).
- **Intersection of Two Planes:** The first two planes intersect along a straight line in 3D space.
- **Equation (iii):** $0x - 3y + 4z = 4$ is a third plane. This third plane cuts through that line at a single intersection point.

### The 3 Geometric Possibilities for 3 Planes:
1. **Unique Solution (Single Common Point):**
   - If all three planes intersect at exactly one single common point, that point $(x, y, z)$ is the unique solution to the system.
2. **Infinite Solutions (A Common Line or Identical Planes):**
   - If the three planes intersect along an entire shared line, or if planes coincide, there are infinitely many common points (infinitely many solutions with free variables).
3. **No Solution (Inconsistent System):**
   - If the three planes do not share any common point (e.g., two planes are parallel, or the three planes form a triangular prism/tunnel where each pair meets in a line but all three never meet simultaneously), no solution exists.



> **Prof. Strang's Observation on the Row Picture in 3D:** While the row picture was easy to visualize in 2D (two lines meeting at a point), visualizing three planes meeting in 3D is already difficult to sketch accurately. In higher dimensions (e.g., 4D, 9D, $n$-D), the row picture becomes almost impossible to visualize, which is why the **Column Picture** is far superior and much easier to understand.

```python
import numpy as np
import plotly.graph_objects as go

# Create a grid of x and y values
x_range = np.linspace(-2, 2, 30)
y_range = np.linspace(-2, 2, 30)
X, Y = np.meshgrid(x_range, y_range)

# Define the equations for the three planes:
# Plane 1: 2x - y = 0 => y = 2x
# To display this properly as a surface, we grid over x and z, computing y
z_range = np.linspace(-1, 2, 30)
X_p1, Z_p1 = np.meshgrid(x_range, z_range)
Y_p1 = 2 * X_p1

# Plane 2: -x + 2y - z = -1 => z = -x + 2y + 1
Z_p2 = -X + 2 * Y + 1

# Plane 3: -3y + 4z = 4 => z = 0.75 * y + 1
Z_p3 = 0.75 * Y + 1

fig = go.Figure()

# Add Plane 1
fig.add_trace(go.Surface(
    x=X_p1, y=Y_p1, z=Z_p1,
    colorscale=[[0, 'blue'], [1, 'blue']],
    opacity=0.5,
    showscale=False,
    name='2x - y = 0'
))

# Add Plane 2
fig.add_trace(go.Surface(
    x=X, y=Y, z=Z_p2,
    colorscale=[[0, 'red'], [1, 'red']],
    opacity=0.5,
    showscale=False,
    name='-x + 2y - z = -1'
))

# Add Plane 3
fig.add_trace(go.Surface(
    x=X, y=Y, z=Z_p3,
    colorscale=[[0, 'orange'], [1, 'orange']],
    opacity=0.5,
    showscale=False,
    name='-3y + 4z = 4'
))

# Highlight the unique intersection point (0, 0, 1)
fig.add_trace(go.Scatter3d(
    x=[0],
    y=[0],
    z=[1],
    mode='markers+text',
    marker=dict(size=10, color='green', symbol='circle', line=dict(color='black', width=2)),
    text=['Unique Solution (0, 0, 1)'],
    textposition='top center',
    name='Unique Solution'
))

# Set title and axes configuration
fig.update_layout(
    title=dict(text='Interactive 3D Row Picture (Rotate for 360-degree View)', font=dict(size=16)),
    scene=dict(
        xaxis=dict(title='X Axis', range=[-2, 2]),
        yaxis=dict(title='Y Axis', range=[-4, 4]),
        zaxis=dict(title='Z Axis', range=[-1, 2]),
        aspectmode='manual',
        aspectratio=dict(x=1, y=1.5, z=1)
    ),
    width=1000,
    height=800,
    margin=dict(l=0, r=0, b=0, t=50)
)

fig.show()
```

## 11. The Column Picture in 3 Dimensions ($\mathbb{R}^3$)

In the **Column Picture**, instead of looking at three intersecting planes, we view the system as a **linear combination of 3-dimensional column vectors**:

$$x \begin{bmatrix} 2 \\ -1 \\ 0 \end{bmatrix} + y \begin{bmatrix} -1 \\ 2 \\ -3 \end{bmatrix} + z \begin{bmatrix} 0 \\ -1 \\ 4 \end{bmatrix} = \begin{bmatrix} 0 \\ -1 \\ 4 \end{bmatrix} \quad \text{--- (Equation i)}$$

Let the column vectors be:
- $\mathbf{c}_1 = \begin{bmatrix} 2 \\ -1 \\ 0 \end{bmatrix}$
- $\mathbf{c}_2 = \begin{bmatrix} -1 \\ 2 \\ -3 \end{bmatrix}$
- $\mathbf{c}_3 = \begin{bmatrix} 0 \\ -1 \\ 4 \end{bmatrix}$
- Target vector: $\mathbf{b} = \begin{bmatrix} 0 \\ -1 \\ 4 \end{bmatrix}$

### Finding the Solution Combination:
Notice that target vector $\mathbf{b}$ is identical to the third column $\mathbf{c}_3$. Therefore, the combination that produces $\mathbf{b}$ is straightforward:
- Take **0** of the first column: $0 \cdot \mathbf{c}_1 = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}$
- Take **0** of the second column: $0 \cdot \mathbf{c}_2 = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}$
- Take **1** of the third column: $1 \cdot \mathbf{c}_3 = \begin{bmatrix} 0 \\ -1 \\ 4 \end{bmatrix}$

$$\mathbf{c}_1' + \mathbf{c}_2' + \mathbf{c}_3' = 0 \begin{bmatrix} 2 \\ -1 \\ 0 \end{bmatrix} + 0 \begin{bmatrix} -1 \\ 2 \\ -3 \end{bmatrix} + 1 \begin{bmatrix} 0 \\ -1 \\ 4 \end{bmatrix} = \begin{bmatrix} 0 \\ -1 \\ 4 \end{bmatrix} = \mathbf{b}$$

Thus, the unique solution to the $3 \times 3$ system is:
$$\mathbf{x = 0, \quad y = 0, \quad z = 1}$$

```python
# 12. Interactive 3D Column Picture (360-degree View using Plotly)
import numpy as np
import plotly.graph_objects as go

# Define vectors
c1 = np.array([2, -1, 0])
c2 = np.array([-1, 2, -3])
c3 = np.array([0, -1, 4])
b = np.array([0, -1, 4])

fig = go.Figure()

# Helper function to draw 3D arrows using Scatter3D
def add_vector(fig, start, end, color, name, width=6):
    # Vector shaft
    fig.add_trace(go.Scatter3d(
        x=[start[0], end[0]], y=[start[1], end[1]], z=[start[2], end[2]],
        mode='lines+markers',
        line=dict(color=color, width=width),
        marker=dict(size=[0, 8], color=color), # arrow-like tip indicator
        name=name
    ))

# Plot columns and the target vector
add_vector(fig, [0,0,0], c1, 'blue', 'c1 = [2, -1, 0]^T')
add_vector(fig, [0,0,0], c2, 'orange', 'c2 = [-1, 2, -3]^T')
add_vector(fig, [0,0,0], c3, 'green', 'c3 = b = [0, -1, 4]^T', width=9)

# Highlight target point b
fig.add_trace(go.Scatter3d(
    x=[b[0]], y=[b[1]], z=[b[2]],
    mode='markers+text',
    marker=dict(size=10, color='green', symbol='circle'),
    text=['b = c3 (0, -1, 4)'],
    textposition='top center',
    name='Target Vector Point'
))

# Configure scene options for uniform aspect ratio and rotation
fig.update_layout(
    title=dict(text='Interactive 3D Column Picture (Rotate for 360-degree View)', font=dict(size=16)),
    scene=dict(
        xaxis=dict(title='X axis', range=[-3, 3]),
        yaxis=dict(title='Y axis', range=[-3, 3]),
        zaxis=dict(title='Z axis', range=[-4, 5]),
        aspectmode='cube'
    ),
    width=1000,
    height=800,
    margin=dict(l=0, r=0, b=0, t=50)
)

fig.show()
```

## 13. The Fundamental Big Question: "Can I solve $Ax = b$ for every $b$?"

Here, $b$ represents **any arbitrary vector in $n$-dimensional space** ($\mathbb{R}^n$).

### Reformulation in Terms of Linear Combinations:
The algebraic question *"Can we solve $Ax = b$ for every $b$?"* is completely equivalent to the geometric question:
> **"Do the linear combinations of the columns of matrix $A$ fill the entire $n$-dimensional space?"**

- For this specific $3 \times 3$ matrix:
  $$A = \begin{bmatrix} 2 & -1 & 0 \\ -1 & 2 & -1 \\ 0 & -3 & 4 \end{bmatrix}$$
  - **Answer: YES.** The linear combinations of its 3 columns fill the entire 3D space ($\mathbb{R}^3$).
  - Any vector $b \in \mathbb{R}^3$ can be expressed as a linear combination of these columns.
  - Therefore, matrix $A$ is a **good matrix**, **non-singular**, and **invertible**.

## 14. When Does the Answer Become "NO"? (The Singular Case)

The system fails to have a solution for every $b$ when the columns do not span the entire space:
1. **Columns Lie in the Same Plane (Coplanar):**
   - If all three 3D column vectors happen to lie on the same flat 2D plane through the origin, all their linear combinations are trapped within that plane.
   - They can only span that 2D plane, not the full 3D space.
   - If target vector $b$ lies outside that plane, it is unreachable: **No solution exists for that $b$**.
2. **Linear Dependence:**
   - This happens when one column contributes nothing new (e.g., if Column 3 is simply Column 1 + Column 2).
   - In that case, the matrix is **singular** and **not invertible**.

### 14.1. The 9-Dimensional Space Intuition (Higher Dimensions)

In the lecture, Professor Strang challenges us to think beyond 3 dimensions and consider a system with 9 equations and 9 unknowns (a $9 \times 9$ matrix $A$):

* **The $9$-Dimensional Vectors:** In this case, we have 9 column vectors, each residing in a 9-dimensional space ($\mathbb{R}^9$).
* **The Random Matrix Scenario:** If you generate a random $9 \times 9$ matrix (for instance, using MATLAB or NumPy), its columns are almost guaranteed to be linearly independent. They will span and completely fill the entire 9-dimensional space, meaning the matrix is invertible, and a unique solution exists for any target vector $\mathbf{b}$.
* **The Failure Case (8D Hyperplane):** However, if the 9th column happens to be identical to the 8th column, the 9th column contributes no new direction. All 9 vectors will be trapped within an 8-dimensional subspace (an **8-dimensional hyperplane**). If your target vector $\mathbf{b}$ lies outside of this 8D hyperplane, no solution to the system exists.

## 15. The Singular Case: When Do Vectors Lie in the Same Plane?

To see clearly when three 3D column vectors fail to span the entire 3D space ($\mathbb{R}^3$), let us examine a specific system where the coefficient matrix $A$ has all zeros in its third row:

$$A = \begin{bmatrix} a_1 & b_1 & d_1 \\ a_2 & b_2 & d_2 \\ a_3 & b_3 & d_3 \end{bmatrix}$$

If the third row is entirely zero, meaning $a_3 = 0, b_3 = 0, d_3 = 0$, the matrix becomes:

$$A = \begin{bmatrix} a_1 & b_1 & d_1 \\ a_2 & b_2 & d_2 \\ 0 & 0 & 0 \end{bmatrix}$$

Splitting matrix $A$ into its three column vectors gives:
$$\mathbf{c}_1 = \begin{bmatrix} a_1 \\ a_2 \\ 0 \end{bmatrix}, \quad \mathbf{c}_2 = \begin{bmatrix} b_1 \\ b_2 \\ 0 \end{bmatrix}, \quad \mathbf{c}_3 = \begin{bmatrix} d_1 \\ d_2 \\ 0 \end{bmatrix}$$

### Geometric Consequence ($z = 0$):
- For each column vector, the third coordinate ($z$-component) is strictly zero.
- This means all three vectors $\mathbf{c}_1, \mathbf{c}_2, \mathbf{c}_3$ lie flat entirely inside the $xy$-plane ($z = 0$).
- Any linear combination of them:
  $$x\mathbf{c}_1 + y\mathbf{c}_2 + z\mathbf{c}_3 = x\begin{bmatrix} a_1 \\ a_2 \\ 0 \end{bmatrix} + y\begin{bmatrix} b_1 \\ b_2 \\ 0 \end{bmatrix} + z\begin{bmatrix} d_1 \\ d_2 \\ 0 \end{bmatrix} = \begin{bmatrix} x a_1 + y b_1 + z d_1 \\ x a_2 + y b_2 + z d_2 \\ 0 \end{bmatrix}$$
  will ALWAYS have a third component of $0$.
- **The Result:** The linear combinations can never produce any target vector $\mathbf{b}$ whose $z$-component is non-zero (for example, $\mathbf{b} = \begin{bmatrix} 1 \\ 2 \\ 5 \end{bmatrix}$ is impossible to reach). The linear combinations are trapped forever on that single 2D plane. Consequently, the matrix is singular and not invertible.

## 16. How to Test Whether Column Vectors Lie on the Same Plane (Linear Independence vs. Dependence)

Three vectors in $\mathbb{R}^3$ lie in the same plane if and only if they are **linearly dependent**.

### 1. The General Test Equation:
Set up the homogeneous linear combination equation equal to the zero vector:

$$x \mathbf{c}_1 + y \mathbf{c}_2 + z \mathbf{c}_3 = \mathbf{0} = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}$$

- **Linearly Dependent (Vectors Lie in the Same Plane):**
  If there exists a solution where **not all scalars $x, y, z$ are zero** (non-trivial solution).
- **Linearly Independent (Vectors Do NOT Lie in the Same Plane):**
  If the **ONLY** way to satisfy the equation is the trivial solution:
  $$x = 0, \quad y = 0, \quad z = 0$$
  When independent, the three columns do not lie on the same plane, and their linear combinations fill the entire 3D space ($\mathbb{R}^3$).

### 2. Dependency in Terms of Direct Linear Combinations:
Vectors are linearly dependent if one vector can be expressed as a combination of the other two:
$$k_1 \mathbf{c}_1 + k_2 \mathbf{c}_2 = \mathbf{c}_3 \quad \text{or} \quad k_2 \mathbf{c}_2 + k_3 \mathbf{c}_3 = \mathbf{c}_1 \quad \text{or} \quad k_1 \mathbf{c}_1 + k_3 \mathbf{c}_3 = \mathbf{c}_2$$

As Prof. Strang emphasized in the lecture: If column 3 is simply column 1 plus column 2 ($\mathbf{c}_3 = \mathbf{c}_1 + \mathbf{c}_2$), it lies in the plane spanned by $\mathbf{c}_1$ and $\mathbf{c}_2$ and contributes no new direction.

## 17. The Core Principle of Planar Vectors

> *If all vectors lie within a single plane, their linear combinations will generate ONLY points that lie strictly within that same plane. They cannot escape the plane to fill the 3-dimensional space.*

- If target vector $\mathbf{b}$ is on that plane $\implies$ **Infinitely many solutions**.
- If target vector $\mathbf{b}$ is off that plane $\implies$ **No solution exists**.

### 17.1. Crucial Distinction: Spanning in 2D vs. 3D and Higher Dimensions

To build a rock-solid intuition, we must understand why lying in "the same plane" is perfectly fine for two vectors in 2D, but becomes a critical failure case for three vectors in 3D, and how this generalizes to higher dimensions:

#### 1. In 2-Dimensional Space ($\mathbb{R}^2$):
* **The Rule:** If you have **two** column vectors that are "unique" (meaning they are not scalar multiples of each other, so they point in different directions), they will **always** span the entire 2D space.
* **Why the Plane is Not a Problem:** In $\mathbb{R}^2$, the entire space is already a single flat plane. Therefore, both vectors lying in the "same plane" is exactly what we want! As long as they do not lie on the exact same *line* (i.e., they are not collinear), their linear combinations will successfully fill and span all of $\mathbb{R}^2$.

#### 2. In 3-Dimensional Space ($\mathbb{R}^3$):
* **The Rule:** If you have **three** column vectors, they must point in three distinct directions to span the entire 3D space.
* **The Plane as a Failure Case:** In $\mathbb{R}^3$, the entire space has depth (it is 3-dimensional). If all three vectors lie in the "same plane" (meaning they are coplanar), they are trapped in a 2D subspace.
* **The Limit:** Because they lie on the same flat sheet, you can stretch, scale, and add them as much as you want, but their sum will never rise off that sheet to fill the 3rd dimension. This is why three coplanar vectors in 3D are **linearly dependent** and fail to solve $Ax = b$ for every $b$.

#### 3. Generalizing to Higher Dimensions (e.g., 3D Subspace in $\mathbb{R}^4$):
* **The Concept:** This geometric limitation scales up identically. Imagine we are in **4-dimensional space** ($\mathbb{R}^4$) and we have **four** column vectors.
* **The Failure Case:** If these four vectors are dependent such that they all lie within a **3-dimensional subspace** (a 3D "hyperplane" nested inside $\mathbb{R}^4$), they are trapped.
* **The Limit:** No matter how you combine, scale, or add these four vectors, their linear combinations will only fill that 3D subspace. They can never escape into the 4th dimension to span the full $\mathbb{R}^4$. Thus, if your target vector $\mathbf{b}$ has any component pointing in that 4th direction, the system $Ax = b$ becomes impossible to solve.
