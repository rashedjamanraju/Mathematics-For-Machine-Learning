## 1. Introduction to Elimination

The method used by every software package to solve linear systems is **elimination**. If the matrix $A$ is a "good" matrix, elimination succeeds and gets the answer efficiently.

Consider a system of 3 equations with 3 unknowns:
$$\begin{aligned}
x + 2y + z &= 2 \quad &\text{--- (Equation 1)} \\
3x + 8y + 1z &= 12 \quad &\text{--- (Equation 2)} \\
0x + 4y + 1z &= 2 \quad &\text{--- (Equation 3)}
\end{aligned}$$

This can be written in matrix form $Ax = b$:
$$A = \begin{pmatrix} 1 & 2 & 1 \\ 3 & 8 & 1 \\ 0 & 4 & 1 \end{pmatrix}, \quad b = \begin{pmatrix} 2 \\ 12 \\ 2 \end{pmatrix}$$

Instead of writing the variables $x, y, z$ repeatedly, we use the matrix $A$ and the augmented matrix $[A|b]$ to perform elimination.

## 2. The Elimination Process: Step-by-Step

The goal of elimination is to transform the matrix $A$ into an upper triangular matrix $U$. This is achieved by systematically knocking out (making zero) the values below the diagonal elements, known as **pivots**.

### Step 1: The First Pivot
The first pivot is the first non-zero element in row 1, which is in the $(1,1)$ position. Here, the first pivot is **1**.
$$A = \begin{pmatrix} \mathbf{[1]} & 2 & 1 \\ 3 & 8 & 1 \\ 0 & 4 & 1 \end{pmatrix}$$

> **Note on Failure:** A pivot cannot be zero. If there is a zero in the pivot position, a row exchange must be performed to bring a non-zero number into that position. If no non-zero number exists below it, the elimination fails completely, and the matrix is singular (not invertible).

We want to eliminate the $x$ term from the second equation. This means making the value in the $(2,1)$ position a zero.
*   **Multiplier:** What number times row 1 should be subtracted from row 2? The multiplier is **3**.
*   **Operation:** Row 2 $\rightarrow$ Row 2 - 3 $\times$ Row 1 ($L_2 \rightarrow L_2 - 3L_1$).

$$A_{\text{updated}} = \begin{pmatrix} 1 & 2 & 1 \\ \mathbf{0} & \mathbf{2} & \mathbf{-2} \\ 0 & 4 & 1 \end{pmatrix}$$
*(Row 2 changed: $(3, 8, 1) - 3 \times (1, 2, 1) = (0, 2, -2)$)*

Next, we look at the $(3,1)$ position. It is already **0**, so no operation is needed for this step.

### Step 2: The Second Pivot
Now, our target is to knock out $y$ from the third equation. The second pivot is the first non-zero element in the updated row 2, in the $(2,2)$ position. The second pivot is **2**.
$$A = \begin{pmatrix} 1 & 2 & 1 \\ 0 & \mathbf{[2]} & -2 \\ 0 & 4 & 1 \end{pmatrix}$$

We want to eliminate the value below it, in the $(3,2)$ position, making it a zero.
*   **Multiplier:** What number times row 2 should be subtracted from row 3? The multiplier is **2**.
*   **Operation:** Row 3 $\rightarrow$ Row 3 - 2 $\times$ Row 2 ($L_3 \rightarrow L_3 - 2L_2$).

$$A_{\text{updated}} = \begin{pmatrix} 1 & 2 & 1 \\ 0 & 2 & -2 \\ \mathbf{0} & \mathbf{0} & \mathbf{5} \end{pmatrix} = U$$
*(Row 3 changed: $(0, 4, 1) - 2 \times (0, 2, -2) = (0, 0, 5)$)*

We have reached the third pivot, which is **5**. There are no rows below it to eliminate, so the process ends. The initial matrix $A$ has been transformed into the upper triangular matrix $U$.

> **Determinant Note:** The determinant of the matrix can be found by multiplying the pivots together. In this case, $1 \times 2 \times 5 = 10$.

### The Augmented Matrix
When performing elimination on $A$, the same operations must be applied to the right-hand side vector $b$. We often represent this using an augmented matrix $[A|b]$.

**Initially:**
$$\begin{pmatrix} 1 & 2 & 1 & \big| & 2 \\ 3 & 8 & 1 & \big| & 12 \\ 0 & 4 & 1 & \big| & 2 \end{pmatrix}$$

**After 1st Elimination ($L_2 \rightarrow L_2 - 3L_1$):**
$$\begin{pmatrix} 1 & 2 & 1 & \big| & 2 \\ 0 & 2 & -2 & \big| & 6 \\ 0 & 4 & 1 & \big| & 2 \end{pmatrix}$$
*(Vector $b$ changed: $12 - 3 \times 2 = 6$)*

**After 2nd Elimination ($L_3 \rightarrow L_3 - 2L_2$):**
$$\begin{pmatrix} 1 & 2 & 1 & \big| & 2 \\ 0 & 2 & -2 & \big| & 6 \\ 0 & 0 & 5 & \big| & -10 \end{pmatrix} = [U|c]$$
*(Vector $b$ changed: $2 - 2 \times 6 = -10$)*

The updated right-hand side vector is called $c$.

## 3. Back Substitution

We now have an upper triangular system of equations $Ux = c$:
$$\begin{aligned}
x + 2y + z &= 2 \quad &\text{--- (i)} \\
2y - 2z &= 6 \quad &\text{--- (ii)} \\
5z &= -10 \quad &\text{--- (iii)}
\end{aligned}$$

Because the system is triangular, we can easily solve it in reverse order, starting from the bottom equation. This is **back substitution**.

1.  **Solve for $z$ from (iii):**
    $$5z = -10 \implies z = -2$$

2.  **Substitute $z$ into (ii) and solve for $y$:**
    $$2y - 2(-2) = 6 \implies 2y + 4 = 6 \implies 2y = 2 \implies y = 1$$

3.  **Substitute $y$ and $z$ into (i) and solve for $x$:**
    $$x + 2(1) + (-2) = 2 \implies x = 2$$

The solution is the vector $x = \begin{pmatrix} 2 \\ 1 \\ -2 \end{pmatrix}$.

```python
import numpy as np
import plotly.graph_objects as go

# Defining the planes for the system Ux = c
# x + 2y + z = 2
# 2y - 2z = 6
# 5z = -10

y_range = np.linspace(-3, 3, 30)
z_range = np.linspace(-4, 0, 30)
Y, Z = np.meshgrid(y_range, z_range)

# Plane 1: x = 2 - 2y - z
X_p1 = 2 - 2 * Y - Z

# Plane 2: y = 3 + z (represented in 3D)
x_range2 = np.linspace(0, 4, 30)
X2, Z2 = np.meshgrid(x_range2, z_range)
Y_p2 = 3 + Z2

# Plane 3: z = -2 (represented in 3D)
x_range3 = np.linspace(0, 4, 30)
y_range3 = np.linspace(0, 4, 30)
X3, Y3 = np.meshgrid(x_range3, y_range3)
Z_p3 = np.full(X3.shape, -2)

fig = go.Figure()

# Add Plane 1 (x + 2y + z = 2)
fig.add_trace(go.Surface(x=X_p1, y=Y, z=Z, colorscale='Blues', opacity=0.5, showscale=False, name='x + 2y + z = 2'))

# Add Plane 2 (2y - 2z = 6)
fig.add_trace(go.Surface(x=X2, y=Y_p2, z=Z2, colorscale='Reds', opacity=0.5, showscale=False, name='2y - 2z = 6'))

# Add Plane 3 (5z = -10)
fig.add_trace(go.Surface(x=X3, y=Y3, z=Z_p3, colorscale='Greens', opacity=0.5, showscale=False, name='5z = -10'))

# Highlight the solution point (2, 1, -2)
fig.add_trace(go.Scatter3d(
    x=[2], y=[1], z=[-2],
    mode='markers+text',
    marker=dict(size=10, color='black', symbol='circle', line=dict(color='white', width=2)),
    text=['Solution (2, 1, -2)'],
    textposition='top center',
    name='Solution Point'
))

fig.update_layout(
    title='Row Picture of the Upper Triangular System Ux = c',
    scene=dict(
        xaxis=dict(title='X Axis'),
        yaxis=dict(title='Y Axis'),
        zaxis=dict(title='Z Axis')
    ),
    width=900, height=700
)
fig.show()
```

## 4. Matrix Multiplication (Rows vs. Columns)

Before defining elimination matrices, let's understand how a matrix operates on rows and columns.

### 1. Matrix $\times$ Column Vector (Linear Combination of Columns)
Multiplying a square matrix by a column vector produces a linear combination of the columns of the matrix.

$$\begin{pmatrix} c_1 & c_2 & c_3 \\ d_1 & d_2 & d_3 \\ f_1 & f_2 & f_3 \end{pmatrix} \begin{pmatrix} a_1 \\ a_2 \\ a_3 \end{pmatrix} = a_1 \begin{pmatrix} c_1 \\ d_1 \\ f_1 \end{pmatrix} + a_2 \begin{pmatrix} c_2 \\ d_2 \\ f_2 \end{pmatrix} + a_3 \begin{pmatrix} c_3 \\ d_3 \\ f_3 \end{pmatrix}$$

The result is one column vector.

### 2. Row Vector $\times$ Matrix (Linear Combination of Rows)
Multiplying a row vector by a square matrix produces a linear combination of the rows of the matrix. This is the crucial operation for elimination.

$$\begin{pmatrix} a_1 & a_2 & a_3 \end{pmatrix} \begin{pmatrix} c_1 & c_2 & c_3 \\ d_1 & d_2 & d_3 \\ f_1 & f_2 & f_3 \end{pmatrix} = a_1 \begin{pmatrix} c_1 & c_2 & c_3 \end{pmatrix} + a_2 \begin{pmatrix} d_1 & d_2 & d_3 \end{pmatrix} + a_3 \begin{pmatrix} f_1 & f_2 & f_3 \end{pmatrix}$$

The result is one row vector.

### Matrix Multiplication Example
Let's see how matrix multiplication combines these ideas.
$$\begin{pmatrix} 1 & 2 & 1 \\ 3 & 8 & 1 \\ 0 & 4 & 1 \end{pmatrix} \begin{pmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \\ 7 & 8 & 9 \end{pmatrix} = \begin{pmatrix} 16 & 20 & 24 \\ 42 & 54 & 66 \\ 23 & 28 & 33 \end{pmatrix}$$

*   **By Columns:** The first column of the result $\begin{pmatrix} 16 \\ 42 \\ 23 \end{pmatrix}$ is formed by taking $1 \times \text{col 1} + 4 \times \text{col 2} + 7 \times \text{col 3}$ of the first matrix.
    $$\begin{pmatrix} 1 & 2 & 1 \\ 3 & 8 & 1 \\ 0 & 4 & 1 \end{pmatrix} \begin{pmatrix} 1 \\ 4 \\ 7 \end{pmatrix} = \begin{pmatrix} 16 \\ 42 \\ 23 \end{pmatrix}$$
*   **By Rows:** The first row of the result $\begin{pmatrix} 16 & 20 & 24 \end{pmatrix}$ is formed by taking $1 \times \text{row 1} + 2 \times \text{row 2} + 1 \times \text{row 3}$ of the second matrix.
    $$\begin{pmatrix} 1 & 2 & 3 \end{pmatrix} \begin{pmatrix} 1 & 2 & 1 \\ 3 & 8 & 1 \\ 0 & 4 & 1 \end{pmatrix} = \begin{pmatrix} 7 & 30 & 6 \end{pmatrix}$$

> **Key Rule:** To perform column operations, multiply by a matrix on the **right**. To perform row operations (like in elimination), multiply by a matrix on the **left**.

## 5. Elimination Matrices

We can express the elimination steps as matrix multiplications using **elementary matrices** (or elimination matrices), denoted as $E_{ij}$, where $i$ and $j$ represent the row and column position being eliminated.

Each elimination step on matrix $A$ is represented as: $\text{Elimination Matrix } \times \text{Updated } A$ (or Initial $A$).

### First Elimination Step: $E_{21}$
We want a matrix that, when multiplied on the left of $A$, subtracts $3 \times L_1$ from $L_2$ to eliminate the $(2,1)$ position. The first and third rows should remain unchanged.

Let's start with the identity matrix, which does nothing:
$$I = \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix}$$

To get our desired operation, we put the negative of the multiplier ($-3$) in the $(2,1)$ position of the identity matrix.

$$E_{21} = \begin{pmatrix} 1 & 0 & 0 \\ -3 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix}$$

Let's verify:
$$E_{21} A = \begin{pmatrix} 1 & 0 & 0 \\ -3 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix} \begin{pmatrix} 1 & 2 & 1 \\ 3 & 8 & 1 \\ 0 & 4 & 1 \end{pmatrix} = \begin{pmatrix} 1 & 2 & 1 \\ 0 & 2 & -2 \\ 0 & 4 & 1 \end{pmatrix}$$

### Second Elimination Step: $E_{32}$
Next, we want to eliminate the $(3,2)$ position of the updated matrix (let's call it $A'$). The step was $L_3 \rightarrow L_3 - 2L_2$.

We put the negative of the multiplier ($-2$) in the $(3,2)$ position of the identity matrix.

$$E_{32} = \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & -2 & 1 \end{pmatrix}$$

Verifying the operation:
$$E_{32} (E_{21}A) = \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & -2 & 1 \end{pmatrix} \begin{pmatrix} 1 & 2 & 1 \\ 0 & 2 & -2 \\ 0 & 4 & 1 \end{pmatrix} = \begin{pmatrix} 1 & 2 & 1 \\ 0 & 2 & -2 \\ 0 & 0 & 5 \end{pmatrix} = U$$

If there were a $(3,1)$ elimination required, it would have been represented by an $E_{31}$ matrix.

## 6. Combining Elimination Steps

The entire elimination process from $A$ to $U$ can be written as:
$$E_{32} (E_{21} A) = U$$

Thanks to the **associative law** of matrix multiplication (we can move parentheses, though we cannot change the order), this is equivalent to:
$$(E_{32} E_{21}) A = U$$

We can define a single matrix $E = E_{32} E_{21}$ that performs all elimination steps at once.
$$E A = U$$

## 7. Permutation Matrices

In some cases, a pivot will be zero, requiring us to swap rows to make that pivot non-zero. This swap operation is expressed by multiplying with a **permutation matrix**, denoted as $P$.

To find the matrix that exchanges rows, perform the exchange on the identity matrix.

For example, to exchange row 1 and row 2 of a $2 \times 2$ matrix $\begin{pmatrix} a & b \\ c & d \end{pmatrix}$:
$$P = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$$
$$P \begin{pmatrix} a & b \\ c & d \end{pmatrix} = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} \begin{pmatrix} a & b \\ c & d \end{pmatrix} = \begin{pmatrix} c & d \\ a & b \end{pmatrix}$$

Remember, multiplying by a permutation matrix on the **left** exchanges rows. Multiplying on the **right** exchanges columns.

## 8. Inverses (A Brief Introduction)

An inverse matrix undoes a given step. The matrix that undoes the elimination step of $E_{21}$ is its inverse, denoted $E_{21}^{-1}$.

The operation for $E_{21}$ was to subtract $3 \times L_1$ from $L_2$. To undo this, we **add** $3 \times L_1$ back to $L_2$.

$$E_{21} = \begin{pmatrix} 1 & 0 & 0 \\ -3 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix}$$

To find the inverse, we change the sign of the off-diagonal element:
$$E_{21}^{-1} = \begin{pmatrix} 1 & 0 & 0 \\ 3 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix}$$

Multiplying a matrix by its inverse results in the Identity matrix $I$:
$$E_{21}^{-1} E_{21} = \begin{pmatrix} 1 & 0 & 0 \\ 3 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix} \begin{pmatrix} 1 & 0 & 0 \\ -3 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix} = \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix} = I$$
